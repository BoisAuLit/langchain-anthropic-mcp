### 🛠️ 第 2 步：定义 Content Planning Agent

</div>

按 João 的最佳实践：`role` 要具体到细分领域，`goal` 要可衡量，`backstory` 写清专长与约束（30–45s、一人在家可拍）。

```python
# Define the Content Planning Agent
content_creator_assistant = Agent(
    role="YouTube Shorts Micro-History Strategist",   # 具体职位 > 泛泛的 "assistant"
    goal="Plan a 1-week slate of high-retention YouTube Shorts about surprising origins of everyday things.",   # 一个明确、可检验的目标
    backstory=(   # 不是装饰：会进 system prompt，决定 LLM 怎么扮演
        "You specialize in 30–45s micro-history that hooks fast, pays off with a twist, and drives comments. "
        "You keep ideas filmable by a solo creator at home with minimal props."
    ),
    llm="gpt-4o-mini",   # 直接写模型名字符串
    verbose=True   # 打印 agent 的思考与输出，学习/调试用
)

print("✓ Content planning agent defined")
```

<div class="dark-title" style="background:linear-gradient(90deg,#1e3a8a,#5b21b6); color:#f1f5f9; padding:6px 20px; border-radius:6px; width:97%;"><style>.dark-title strong{color:#fde047;}</style>

### 🛠️ 第 3 步：定义 Content Planning Task

</div>

`description` 给平台、细分方向、4 个目标和创作约束；`expected_output` 直接写出 JSON schema，LLM 会照着格式输出。

```python
# Define the Content Planning Task
task = Task(
    description=(    # 做什么 + 上下文 + 约束
        "Create a 1-week video posting plan with 5 video blueprints. "
        "Platform: YouTube Shorts (vertical 9:16, 30-45s). "
        "Niche: Micro-History of Everyday Things (e.g., why pencils are yellow, origins of bubble wrap, etc.). "
        "Primary goals: 1) thumb-stop hook in first 1s, 2) crystal-clear narrative with a surprise, "
        "3) strong SEO phrasing in title/caption, 4) comment-bait CTA. "
        "Context: solo creator, home-filmable, no special gear. "
    ),
    expected_output=(   # 结果的形状：这里给出 JSON schema
        '''
        Output a JSON array following the schema below, which contains a
        weekly schedule and 5 video blueprints. Each video blueprint should include:
        {
          "videos": [
            {
              "title": "<searchable, curiosity-driven title>",
              "hook_main": "<<=12 words, shows payoff fast>",
              "hook_alt": "<variant hook>",
              "visuals": ["simple prop or b-roll idea 1", "idea 2"],
              "tags": ["#microhistory","#everydaythings","#shorts"],
              "cta": "<question that invites comments>"
            }
          ]
        }
        '''
    ),
    agent=content_creator_assistant   # 绑定执行者
)

print("✓ Content planning task defined")
```

**`expected_output` 只是「提示」，不是强制**：模型通常会照做，但不保证是合法 JSON。要强制结构化输出需要 `output_pydantic` / `output_json`（后面模块会讲）。

### 🛠️ 第 4 步：组 Crew 并运行

</div>

<div style="background:#111827; border:1px dashed #334155; border-radius:8px; padding:10px 16px; width:97%;"><div style="color:#94a3b8; font-weight:700; margin-bottom:6px;">单 agent、单 task：跑一次就结束</div><span style="display:inline-block; background:#0f172a; border:1.5px solid #a78bfa; color:#e2e8f0; padding:4px 12px; border-radius:999px; margin:3px 0; font-size:0.9em;"><code style="background:#0b1220; color:#fde68a; padding:1px 5px; border-radius:4px; font-size:0.92em;">kickoff_async()</code></span><span style="color:#fbbf24; margin:0 6px; font-weight:700;">➜</span><span style="display:inline-block; background:#0f172a; border:1.5px solid #60a5fa; color:#e2e8f0; padding:4px 12px; border-radius:999px; margin:3px 0; font-size:0.9em;">按 <code style="background:#0b1220; color:#fde68a; padding:1px 5px; border-radius:4px; font-size:0.92em;">tasks</code> 顺序执行</span><span style="color:#fbbf24; margin:0 6px; font-weight:700;">➜</span><span style="display:inline-block; background:#0f172a; border:1.5px solid #fbbf24; color:#e2e8f0; padding:4px 12px; border-radius:999px; margin:3px 0; font-size:0.9em;">agent 调 LLM 生成</span><span style="color:#fbbf24; margin:0 6px; font-weight:700;">➜</span><span style="display:inline-block; background:#0f172a; border:1.5px solid #4ade80; color:#e2e8f0; padding:4px 12px; border-radius:999px; margin:3px 0; font-size:0.9em;"><code style="background:#0b1220; color:#fde68a; padding:1px 5px; border-radius:4px; font-size:0.92em;">CrewOutput</code></span></div>

<div class="dark-warning" style="background:#2a2418; color:#fde68a; padding:12px 24px; border-left:4px solid #fbbf24; border-radius:4px; width:97%;"><style>.dark-warning strong{color:#f9a8d4;}</style>

**🆕 2026-10 版本差异（crewai 1.15.22）**：在 Jupyter 里用同步的 `crew.kickoff()` 会报
`RuntimeError: Agent execution was invoked synchronously from within a running event loop`——Jupyter 自己已经有一个运行中的事件循环，而新版默认的 `AgentExecutor` 一检测到运行中的循环就改走 async 路径。
所以 notebook 里用 `await crew.kickoff_async()`：它内部是 `asyncio.to_thread(self.kickoff, ...)`，在一个**没有事件循环的工作线程**里跑同步 kickoff。纯 `.py` 脚本里 `crew.kickoff()` 照常可用。

</div>

```python
# Create the content planning crew
crew = Crew(
    agents=[content_creator_assistant],   # 团队名单
    tasks=[task]   # 执行顺序 = 这个列表的顺序
)

print("✓ Content planning crew created")
print("\n🚀 Starting content planning...\n")

# Execute the crew's task
result = await crew.kickoff_async()   # Jupyter 里用 async 版本（见上方 🆕）

print("\n✓ Content planning complete!")
```
