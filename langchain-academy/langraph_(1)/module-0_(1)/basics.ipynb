```python
import os, getpass
from dotenv import load_dotenv, find_dotenv

# load variables from a .env file (searches the current and parent directories)
load_dotenv(find_dotenv())


def _set_env(var: str):
    if not os.environ.get(var):
        os.environ[var] = getpass.getpass(f"{var}: ")


_set_env("OPENAI_API_KEY")
```

**🎯 Chat Model 基础**

LangChain 的 chat model 有一堆[默认方法](https://reference.langchain.com/python/langchain_core/runnables)，最常用两个：

- [`stream`](https://docs.langchain.com/oss/python/langchain/models#stream)：流式返回响应分片
- [`invoke`](https://docs.langchain.com/oss/python/langchain/models#invoke)：一次性调用模型拿到完整响应

Chat model 的输入是 [messages](https://docs.langchain.com/oss/python/langchain/messages)：每条消息有一个 role（谁在说话）和 content（说了什么）。

</div>

```python
from langchain_core.messages import HumanMessage

# Create a message
msg = HumanMessage(content="Hello world", name="Lance")

# Message list
messages = [msg]

# Invoke the model with a list of messages 
gpt4o_chat.invoke(messages)
```

模型返回的是 `AIMessage`。另外，`invoke` 也可以直接传字符串——字符串会被自动包装成 `HumanMessage` 再丢给底层模型，不用每次都手动构造 message 对象。

```python
gpt4o_chat.invoke("hello world")
```

```python
gpt35_chat.invoke("hello world")
```
